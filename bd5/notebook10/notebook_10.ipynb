## 8. Conclusiones

Completa:

- **Topic:** Es el canal donde kafka guarda la informacion
- **Partición:** Son las divisiones que se realizan sobre topic
- **Consumer Group:** Es un grupo de consumidores que leen datos para repartirlos en en sus particiones
- **Offset:** Sirve para saber hasta donde ha leido un consumidor
- **Rebalanceo:** Es cono redistribuye kafka los consumers en los diferentes topic
- **Paralelismo:** La posibilidad de trabajar de manera paralela, es dicir a la vez
- **Escalabilidad:** Es la facilidad que puede tener para aumentar  un proyecto
